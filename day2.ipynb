# 1. Frequency Counting using Dictionaries

Frequency counting is the process of determining how many times each element appears in a collection.

### Example

Input:

```python
arr = [2, 3, 2, 5, 3, 2]
```

Output:

- 2 → 3
- 3 → 2
- 5 → 1

Python dictionaries make frequency counting efficient by storing data as **key-value pairs**.

### Time Complexity
- Average: **O(n)**

### Space Complexity
- **O(n)**

```python
arr = [2, 3, 2, 5, 3, 2]

freq = {}

for num in arr:
    freq[num] = freq.get(num, 0) + 1

print(freq)
```

# 2. Introduction to Hashing

Hashing is a technique used to store and retrieve data efficiently.

Instead of searching every element one by one, a **hash function** maps a key directly to a memory location.

Python dictionaries internally use **hash tables**.

### Advantages
- Fast Search
- Fast Insertion
- Fast Deletion

Average Time Complexity:

- Search → **O(1)**
- Insert → **O(1)**
- Delete → **O(1)**

Worst Case:

- **O(n)** (due to collisions)

## Collision

A collision occurs when two different keys generate the same hash value.

Python automatically resolves collisions internally, allowing dictionaries to continue functioning efficiently.

# 3. Introduction to Recursion

Recursion is a programming technique where a function calls itself to solve a smaller version of the same problem.

Every recursive function must have:

- **Base Case** – Stops recursion.
- **Recursive Call** – Calls the function again with a smaller input.

```python
def recurse(n):
    if n == 0:
        return
    recurse(n - 1)
```

## Call Stack

Every recursive call is stored in the **Call Stack**.

The stack grows until the base case is reached.

After reaching the base case, the function calls return one by one in reverse order.

# 4. Recursion using Parameters

Instead of using global variables, values can be passed through function parameters.

This approach:
- Produces cleaner code

# 5. Functional Recursion

In Functional Recursion, the recursive function **returns a value** instead of directly printing it.

Examples include:
- Factorial
- Sum of N Numbers
- Fibonacci
- Power Calculation

The final answer is built while the recursive calls return.

## 6.1 Factorial (Recursive)

Factorial of `n` is the product of all positive integers ≤ `n`.

- **Base Case**: `n == 0` or `n == 1` → return 1
- **Recursive Call**: `n * factorial(n - 1)`

### Time Complexity
- **O(n)**

### Space Complexity
- **O(n)** (call stack depth)

```python
class Solution:
    def factorial(self, n: int) -> int:
        if n == 0 or n == 1:
            return 1        
        return n * self.factorial (n-1)
```

## 6.2 Palindrome Check — Without Recursion

Checks whether a string reads the same forwards and backwards using the **two-pointer** technique.

- Start pointers at both ends (`left`, `right`) and move inward.
- Return `False` (0) as soon as a mismatch is found; otherwise return `True` (1).

### Time Complexity
- **O(n)**

### Space Complexity
- **O(1)**

```python
# without recursion
class Solution:
    def isPalindrome(self, s: str) -> bool:
        left = 0
        n = len(str(s))
        right = n - 1
        while left < right:
            if s[left] != s[right]:
                return 0
            left += 1
            right -= 1
        return 1
```

## 6.3 Palindrome Check — With Recursion

Same problem solved recursively.

- **Base Case**: `left > right` → the string is a palindrome, return `"True"`.
- If characters at `left` and `right` don't match → return `"False"`.
- **Recursive Call**: move both pointers inward — `func(s, left + 1, right - 1)`.

### Time Complexity
- **O(n)**

### Space Complexity
- **O(n)** (call stack depth)

```python
#with recursion
def func(s,left,right):
    if left > right:
        return "True"
    if s[left] != s[right]:
        return "False"
    return func(s,left+1,right-1)
```

## 6.4 Fibonacci — With Recursion

Classic recursive Fibonacci, where each term is the sum of the two preceding terms.

- **Base Case**: `n == 0` or `n == 1` → return `n`.
- **Recursive Call**: `func(n-1) + func(n-2)`.

### Time Complexity
- **O(2^n)** — exponential, since the same sub-problems are recomputed repeatedly.

### Space Complexity
- **O(n)** (maximum call stack depth)

```python
#with recursion
def func(n):
    if n ==0 or n ==1:
        return n
    return func(n-1) + func(n-2)
# TC - O(2^n) and SC - O(n)
```

## 6.5 Fibonacci — Without Recursion (Iterative)

An iterative approach avoids the exponential blowup of naive recursion by keeping only the last two values in memory.

### Time Complexity
- **O(n)**

### Space Complexity
- **O(1)**

```python
#without recursion
def fib(n):
    a = 0
    b = 1
    for i in range(n):
        print(a,end=" ")
        c = a + b
        a = b
        b = c
fib(4)
# tc - o(n) and sc - o(1)
```

# 📝 Key Takeaways

- Dictionaries use **Hash Tables** for efficient storage.
- Hashing provides **O(1)** average lookup, insertion, and deletion.
- Collisions occur when multiple keys map to the same location.
- Every recursive function requires a **Base Case** and a **Recursive Call**.
- Recursive calls are managed using the **Call Stack**.
- Parameters can be used to pass information through recursive calls.
- Functional recursion returns values while unwinding the recursion stack.
- Naive recursive Fibonacci is exponential (**O(2^n)**); the iterative version is linear (**O(n)**) and more space-efficient.
- The same problem (e.g. palindrome check, Fibonacci) can often be solved both recursively and iteratively — recursion tends to be more elegant, iteration tends to be more space-efficient.
