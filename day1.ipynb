## extracting digits of a given number

```python
n = 5789
num = n
while num>0:
    last_d = num % 10
    print(last_d)
    num = num // 10
```

## count the digits

```python
n = 5789
num = n
count = 0
while num>0:
    last_d = num % 10
    count += 1
    num = num // 10
print(count)
```

## reverse the digits

```python
n= 5789
num = n
rev = 0
while num>0:
    last_d = num % 10
    rev = rev * 10 + last_d
    num = num // 10
print(rev)
```

## palindrome number

```python
n = 131
num = n
rev = 0
while num > 0:
    last_d = num % 10
    rev = rev * 10 + last_d
    num = num // 10
if (rev == n):
    print("the number is a palindrome")
else:
    print("the number is not palindrome")
```

## armstrong number

```python
n = 153
num = n
nod = len(str(n))
total = 0
while num>0:
    last_d = num % 10
    total += last_d ** nod
    num = num // 10
if(total == n):
    print("the number is armstrong")
else:
    print("the number is not armstrong")
```

### printing factors of a number - brute force

```python
n = 12
num = n
result = []
for i in range(1,num+1):
    if (num%i==0):
        result.append(i)
    else:
        continue
print(result)
```

### optimal solution - 1

```python

n = 12
num = n
result = []
for i in range(1,int(num/2)+1):
    if (num%i == 0):
        result.append(i)
    else:
        continue
result.append(num)
print(result)
```

### optimal solution - 2

```python
from math import sqrt
n = 12
num = n
result = []
for i in range(1,int(sqrt(num))+1):
    if(num % i == 0):
        result.append(i)
        if (num // i != i):
            result.append(num//i)
result.sort()
print(result)
```
