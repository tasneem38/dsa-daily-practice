# 1. Extracting Digits of a Number

To pull out each digit of a number, repeatedly use `% 10` to get the last digit, then `// 10` to remove it. This continues until the number becomes 0.

### Time Complexity
- **O(d)**, where `d` is the number of digits

### Space Complexity
- **O(1)**

```python
n = 5789
num = n
while num>0:
    last_d = num % 10
    print(last_d)
    num = num // 10
```

# 2. Count the Digits

Same digit-extraction loop as above, but instead of printing each digit, we increment a counter each time.

### Time Complexity
- **O(d)**

### Space Complexity
- **O(1)**

```python
n = 5789
num = n
count = 0
while num>0:
    last_d = num % 10
    count += 1
    num = num // 10
print(count)
```

# 3. Reverse the Digits

Build the reversed number digit by digit: each extracted digit is appended to `rev` by shifting existing digits left (`rev * 10`) and adding the new digit.

### Time Complexity
- **O(d)**

### Space Complexity
- **O(1)**

```python
n= 5789
num = n
rev = 0
while num>0:
    last_d = num % 10
    rev = rev * 10 + last_d
    num = num // 10
print(rev)
```

# 4. Palindrome Number

A number is a palindrome if it reads the same forwards and backwards. Reuses the digit-reversal logic from above, then compares the reversed number to the original.

### Time Complexity
- **O(d)**

### Space Complexity
- **O(1)**

```python
n = 131
num = n
rev = 0
while num > 0:
    last_d = num % 10
    rev = rev * 10 + last_d
    num = num // 10
if (rev == n):
    print("the number is a palindrome")
else:
    print("the number is not palindrome")
```

# 5. Armstrong Number

A number is an Armstrong number if the sum of its digits, each raised to the power of the total digit count (`nod`), equals the original number.

Example: `153 = 1³ + 5³ + 3³`

### Time Complexity
- **O(d)**

### Space Complexity
- **O(1)**

```python
n = 153
num = n
nod = len(str(n))
total = 0
while num>0:
    last_d = num % 10
    total += last_d ** nod
    num = num // 10
if(total == n):
    print("the number is armstrong")
else:
    print("the number is not armstrong")
```

# 6. Finding Factors of a Number

Three approaches to finding all factors of `n`, from brute force to fully optimized.

## 6.1 Brute Force

Check every number from `1` to `n` and see if it divides `n` evenly.

### Time Complexity
- **O(N)**

### Space Complexity
- **O(N)** (to store the factors, worst case)

```python
n = 12
num = n
result = []
for i in range(1,num+1):
    if (num%i==0):
        result.append(i)
    else:
        continue
print(result)
```

## 6.2 Optimal Solution 1

Key insight: no factor of `n` (other than `n` itself) can be greater than `n / 2`. So we only need to loop up to `n // 2`, then append `n` separately at the end.

### Time Complexity
- **O(N/2)** → simplifies to **O(N)**

### Space Complexity
- **O(N)** (to store the factors, worst case)

```python

n = 12
num = n
result = []
for i in range(1,int(num/2)+1):
    if (num%i == 0):
        result.append(i)
    else:
        continue
result.append(num)
print(result)
```

## 6.3 Optimal Solution 2 (√N Approach)

Factors come in **pairs**: if `i` divides `n`, then `n // i` is also a factor. So we only need to check up to `√n` — for every divisor found, both `i` and `n // i` are added (avoiding duplicates when `i == n // i`).

### Time Complexity
- **O(√N)**

### Space Complexity
- **O(N)** (to store the factors, worst case)

```python
from math import sqrt
n = 12
num = n
result = []
for i in range(1,int(sqrt(num))+1):
    if(num % i == 0):
        result.append(i)
        if (num // i != i):
            result.append(num//i)
result.sort()
print(result)
```

# 📝 Key Takeaways

- `%` extracts the last digit; `//` removes it — together they let you process a number digit by digit.
- Reversing a number and checking palindromes rely on the same core digit-extraction loop.
- Armstrong numbers combine digit extraction with exponentiation based on digit count.
- Factor-finding has three tiers of efficiency:
  - Brute force: **O(N)**
  - Halved range: **O(N/2)** (still effectively O(N))
  - √N approach: **O(√N)** — the most efficient, by exploiting factor pairs.
- Choosing the right loop bound (N, N/2, or √N) is a recurring optimization pattern for number-theory problems.
