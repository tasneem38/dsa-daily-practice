```python
class Solution:
    def factorial(self, n: int) -> int:
        if n == 0 or n == 1:
            return 1        
        return n * self.factorial (n-1)
```

```python
# without recursion
class Solution:
    def isPalindrome(self, s: str) -> bool:
        left = 0
        n = len(str(s))
        right = n - 1
        while left < right:
            if s[left] != s[right]:
                return 0
            left += 1
            right -= 1
        return 1
```

```python
#with recursion
def func(s,left,right):
    if left > right:
        return "True"
    if s[left] != s[right]:
        return "False"
    return func(s,left+1,right-1)
```

```python
#with recursion
def func(n):
    if n ==0 or n ==1:
        return n
    return func(n-1) + func(n-2)
# TC - O(2^n) and SC - O(n)
```

```python
#without recursion
def fib(n):
    a = 0
    b = 1
    for i in range(n):
        print(a,end=" ")
        c = a + b
        a = b
        b = c
fib(4)
# tc - o(n) and sc - o(1)
```
